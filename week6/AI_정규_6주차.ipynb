```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
apple = fruits[0:100].reshape(-1, 100*100)
pineapple = fruits[100:200].reshape(-1, 100*100)
banana = fruits[200:300].reshape(-1,100*100)
```

```python
print(apple.shape)
```

```python
print(apple.mean(axis=1))
```

```python
plt.hist(apple.mean(axis=1), alpha=0.8, label='apple')
plt.hist(pineapple.mean(axis=1), alpha=0.8, label='pineapple')
plt.hist(banana.mean(axis=1), alpha=0.8, label='banana')
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].bar(range(10000), apple.mean(axis=0))
axs[1].bar(range(10000), pineapple.mean(axis=0))
axs[2].bar(range(10000), banana.mean(axis=0))
plt.show()
```

```python
plt.show()
apple_mean = apple.mean(axis=0).reshape(100, 100)
pineapple_mean = pineapple.mean(axis=0).reshape(100, 100)
banana_mean = banana.mean(axis=0).reshape(100,100)

fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(apple_mean, cmap='gray_r')
axs[1].imshow(pineapple_mean, cmap='gray_r')
axs[2].imshow(banana_mean, cmap='gray_r')
plt.show()
```

```python
abs_diff = np.abs(fruits-apple_mean)
abs_mean = np.mean(abs_diff, axis=(1,2))
print(abs_mean.shape)
```

```python
apple_index = np.argsort(abs_mean)[:100]
apple_index = apple_index.reshape(10, 10)
fig, axs = plt.subplots(10, 10, figsize=(10, 10))
for i in range(10):
    for j in range(10):
        axs[i, j].imshow(fruits[apple_index[i, j]], cmap='gray_r')
        axs[i, j].axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt
def draw_fruits(arr, ratio=1):
  n = len(arr)
  rows = int(np.ceil(n/10))
  cols = n if rows < 2 else 10
  fig, axs = plt.subplots(rows, cols, figsize=(cols*ratio, rows*ratio), squeeze=False)
  for i in range(rows):
    for j in range(cols):
      if i*10+j < n:
        axs[i, j].imshow(arr[i*10+j], cmap='gray_r')
      axs[i, j].axis('off')
  plt.show()
```

```python
draw_fruits(fruits[km.labels_==0])
```

```python
draw_fruits(fruits[km.labels_==1])
```

```python
draw_fruits(fruits[km.labels_==2])
```

```python
draw_fruits(km.cluster_centers_.reshape(-1, 100, 100), ratio=3)
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()

data = iris.data
feature_names = iris.feature_names

print(data[:5])
print(feature_names)
```

```python
X = data[:, 2:4]
print(X.shape)
print("사용 변수:", feature_names[2], "/", feature_names[3])
```

```python
inertia = []
k_range = range(2, 7)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertia.append(km.inertia_)
    print(f"k={k} → inertia: {km.inertia_:.2f}")
```

```python
plt.figure(figsize=(8, 4))
plt.plot(k_range, inertia, marker='o')
plt.xlabel('k (군집 수)')
plt.ylabel('Inertia')
plt.title('Elbow Method - 최적 k 찾기')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

```python
best_k = 3
```

```python
km = KMeans(n_clusters=best_k, random_state=42)
km.fit(X)

print("반복 횟수:", km.n_iter_)
print("군집 레이블:", km.labels_)
```

```python
labels, counts = np.unique(km.labels_, return_counts=True)
for label, count in zip(labels, counts):
    print(f"군집 {label}: {count}개")
```

```python
colors = ['tomato', 'steelblue', 'seagreen']
markers = ['o', 's', '^']
```

```python
plt.figure(figsize=(8, 6))

for i in range(best_k):
    mask = km.labels_ == i
    plt.scatter(
        X[mask, 0], X[mask, 1],
        c=colors[i], marker=markers[i],
        label=f'군집 {i} ({counts[i]}개)',
        alpha=0.7, edgecolors='k', linewidths=0.5
    )

# 군집 중심점 표시
centers = km.cluster_centers_
plt.scatter(
    centers[:, 0], centers[:, 1],
    c='gold', marker='*', s=300,
    edgecolors='k', linewidths=1,
    label='중심점 (Centroid)', zorder=5
)

plt.xlabel(feature_names[2])
plt.ylabel(feature_names[3])
plt.title(f'K-평균 군집 결과 (k={best_k})')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
